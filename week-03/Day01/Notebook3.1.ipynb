## Step 2 - Train/Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

## Step 4 - Why the model must never see the test set during training?

When the test set is used during training the model it will produce highly accurate results that are considered misleading. That is because the model records or memorises the results. Consequently, it will produce incorrect results when dealing with new datasets since no pattern recognition mechanism has been developed by the model.
