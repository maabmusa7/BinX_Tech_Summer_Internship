## **What is NLP?**

Natural Language is normally unstructured, noisy and incostitent. Natural Language Preprocessing (NLP) is a branch of AI that  **converts data into clean, structured and standardized format, enabling effective feature extraction that improves the model's performance**. It does so by *turning text into numeric representations* - e.g. tokenization - that models can process. NLP produces human-readable language as well. Simply, NLP *teaches machines hoe to work with human language*.


## **Where is NLP applied?**

NLP is useful for text classification, sentiment analysis, translation, Chatbots and search engines.

## **Text Preprocessing Techniques in NLP**

### **Tokenization**

Tokenization is the process of **breaking text into smaller meaningful units**, such as words or sequences. This converts text onto a structured format that NLP models can analyze and process. This process **preserves the sequence of words** for context in anlysis and prepares thet text for further steps like stopword removal, stemming, etc.

### **Stop Words**

Stop Words are **common words in language that do not add significant meaning** for NLP tasks. Stop Words include articles, prepositions, conjunctions, and auxiliary verbs such as; *is, the, and, a, in, on, of, etc*. Eliminating them helps in **reducing the size of the text data, improving computational efficiency**.

### **Stemming**

Stemming is the process of **reducing words to their root/base form y removing suffuxes**. It helps normalizing text. For example, *"ate"*, *"eating"* are treates as *"eat"* after stemming. Stemming is commonly used because it is **computationally efficient**, improving NLP models performance

### **Lemmatization**

Lemmatization is the process of **converting a word to its meaningful base or dictionary form, called a lemma**. It differs from stemming in *ensuring that the root word is an actual word in the language*, while stemming uses a rule-based approach that doesn't always produce real words. Lemmatization relies on vocabulary databases like *WordNet* to identify the correct base form based on the word part of speach *(POS)*. It is, however, **computationally expensive**.

### **Part-of-Speech (POS) Tagging**

Part-of-Speech (POS) Tagging is the process of **assigning a gramatical category to each word in a sentence based on its meaning and context**. Catgories inlude, noun, verb, adjective, pronoun, preposition, conjunction, interjection, etc. POS helps in **understanding the syntatic structure of a sentence** which is essential for various NLP tasks like; text parsing, machine translation, speech recognition, and sentiment analysis. POS plays a critical role in *context-aware text processing*.


### **Named Entity Recognitioin (NER)**

NER is an NLP technique that **identifies and classifies names and specific entites in text**. It helps extraxt structured information from unstructured text making it useful for various applications.

## **About the Notebook**

This notebook **builds an NLP pipeline** for the *AG News dataset*. The pipeline applies  five steps in order; *lowercasing, punctuation removal, tokenization, stop word removal and lemmatization*.

Since AG News is a **topic classification** task, the main risk in cleaning is nouns/verbs that signal topic; company names, sports terms, tech vocabulary, not losing negation words - which would be a concern for sentiment analysis.

```python
from google.colab import drive

drive.mount('/content/drive')

train_path="/content/drive/MyDrive/train.csv"


df = pd.read_csv(train_path, header=0, names=["label", "title", "description"])
df["text"] = df["title"] + " " + df["description"]


print("Extraction complete!")
```

```python
sample = df["text"].iloc[0]
print("Raw sample:", sample)
```

## **Tokenizing a text sample**

```python
tokens = word_tokenize(sample)
print("Tokens:", tokens)
```

## **Full Cleaning Pipeline**

```python
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

def clean_text(text):
  # lowercase
  text= text.lower()
  # remove punctuation
  text= text.translate(str.maketrans('', '', string.punctuation))
  # tokenize
  tokens = word_tokenize(text)
  # remove stopwords
  tokens = [word for word in tokens if word not in stop_words]
  # lemmatize
  tokens= [lemmatizer.lemmatize(word) for word in tokens]

  return tokens

cleaned_sample = clean_text(sample)
print("Cleaned:", cleaned_sample)

df["cleaned_tokens"] = df["text"].apply(clean_text)
```

## **Verifying the cleaning preserves task-critical words**

```python
print(sorted(stop_words)[:30])

critical_words = ["not", "no", "against", "up", "down", "won", "lost"]
for w in critical_words:
  print(w,"-> stop word?", w in stop_words)
```

## Findings & Observations

**Stopword list check:**
critical words that could carry topic signal ('not', 'no', 'against', 'up', 'down', 'won', 'lost') compared to the standard NLTK English stopword list showed that all word listed above are in default stopword list except 'lost'.

Since this is a topic classification task, removing negation words like 'not' and 'no' wouldn't downgrade the model's ability to distinguish topics.

Removing words like 'down' and 'won' that point to certain topics could have a considerable impact on the model. However, the stopword list will remain unmodified since other remaining nouns carried sufficient topic signal on their own (team names, scores, company names).

**Limitation:** Punctuation removal before tokenization lead to some words being merged together such as "dwindling\band" -> "dwindlingband". This has minimal impact on topic classification for that reason no furthor adjustments were done.
