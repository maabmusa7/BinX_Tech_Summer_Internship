# Day 1 — Descriptive Statistics 

## Step 1:

```python
import numpy as np
import pandas as pd
import statistics as stats
import matplotlib.pyplot as plt
```

```python
taxis = pd.read_csv("taxis.csv")
distance = np.array(taxis["trip_distance"])
print(distance)
print("Total Number of Trips: ", distance.shape)
```

- Statistics, Matplorlib, Pandas and NumPy imported.
- Dataset was converted to a DataFrame using .read_csv()(Pandas).
- "trip_distance" column was extracted and converted into a NumPy array.

## Step 2:

### Measures of center

```python
mean = np.mean(distance)
median = np.median(distance)
mode = stats.mode(distance)

print("Measures of center for trip distances:")
print("Mean (Average) distance:", mean, "mi")
print("Median (Middle value):",median, "mi")
print("Mode (most repeated value):", mode, "mi")
```

### Measures of variability

```python
std = np.std(distance)
q1, q3 =np.percentile(distance, [25,75])
iqr = q3 - q1

print("Measures of variability/spread for trip distances:")
print("Standard deviation:", std, "mi")
print("Inter quartile range [IQR]:", iqr, "mi")
```

```python
print("Median of upper half of the data:", q3)
print("Median of lower half of the data:", q1)
print("The middle 50% of trips(IQR) fall between", q1,"&", q3,"miles.") 
```

```python
range = max(distance)-min(distance)
print("Maximum trip distance:", max(distance), "mi")
print("Minimum trip distance:", min(distance),"mi")
print("Range of trip distances:", range, "mi")
```

```python
variance = stats.variance(distance)
print("The average squared deviation from the mean (Variance):", variance, "mi")
```

## Step 3:

```python
trips_above_mean = (distance > mean).mean()*100
trips_above_median = (distance > median).mean()*100

print(f"{trips_above_mean:.1f}%", "of trips are above the Average.")
print(f"{trips_above_median:.1f}%", "of trips are above the Median.")#f"{ :.1f}%" used to round to 1 decimal place
```

* Median best represents a typical trip. Only 26.5% of the trips are longer than than the average distance while 49.9% of the trips are above the average trip distance.
* The median splits the data exactly in half which makes it more valid.

```python
plt.figure(figsize=(6,4))
plt.hist(taxis["trip_distance"].to_numpy(), bins = 10, color="#4C72B0", edgecolor="white")
plt.axvline(mean, color = "red", linestyle="--", label=f"Mean:{mean:.2f} mi")
plt.axvline(median, color = "green", linestyle="--", label=f"Median:{median:.2f} mi")
plt.title("Distribution of Taxi Trip Distances")
plt.xlabel("Distancs in Miles")
plt.ylabel("Number of Trips")
plt.show()
```

## Step 4:

- Taxi trip distances in the data set center around a median of 1.64 miles; half of all trips were shorter than 1.64 miles.
- The middle 50% of trips(IQR) fall between 0.98 & 3.2225 miles.
That means the typical ride is short, approximately 2 miles.
- The mean(3.05) is nearly double the meadian(1.64) which is a sign for right skew.

- Most trip distances concentrate around 3 miles (mean).
- Skewed data is best rpresented by median and IQR bacause the standard deviation and mean are inflated by outliers.
